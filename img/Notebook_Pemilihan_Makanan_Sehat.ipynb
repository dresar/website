## Penjelasan Sederhana Random Forest

Random Forest adalah algoritma pembelajaran mesin yang sangat kuat dan serbaguna, terutama untuk tugas klasifikasi dan regresi. Nama 'Random Forest' sendiri memberikan petunjuk tentang cara kerjanya: ini adalah 'hutan' yang terdiri dari banyak 'pohon' acak.

### 1. Konsep Dasar: Hutan Pohon Keputusan

Bayangkan Anda memiliki beberapa teman yang ahli dalam memprediksi apakah suatu makanan sehat atau tidak sehat. Setiap teman ini (kita sebut saja 'pohon keputusan') akan melihat beberapa ciri makanan (seperti jumlah gula, lemak jenuh, serat, dll.) dan kemudian membuat keputusan sendiri. Misalnya, satu teman mungkin sangat memperhatikan gula, sementara yang lain lebih fokus pada serat dan sodium.

Random Forest bekerja dengan membangun banyak (biasanya ratusan atau ribuan) pohon keputusan yang berbeda secara independen. Setiap pohon dilatih menggunakan sebagian kecil data pelatihan yang dipilih secara acak (disebut bootstrapping) dan hanya mempertimbangkan sebagian kecil fitur secara acak pada setiap 'cabang' keputusan. Keragaman ini membuat setiap pohon memiliki perspektif yang sedikit berbeda tentang data.

### 2. Mengapa Random Forest Lebih Baik dari Satu Pohon Keputusan?

Satu pohon keputusan tunggal cenderung 'overfit' pada data pelatihan. Artinya, pohon tersebut mungkin terlalu spesifik mengingat detail kecil dari data pelatihan sehingga tidak dapat menggeneralisasi dengan baik ke data baru yang belum pernah dilihat sebelumnya. Ini seperti seorang teman yang menghafal semua jawaban ujian tanpa benar-benar memahami konsepnya.

Random Forest mengatasi masalah ini dengan menggabungkan hasil dari banyak pohon. Ini mirip dengan meminta pendapat dari banyak ahli yang berbeda. Jika satu ahli membuat kesalahan, kesalahan tersebut kemungkinan akan diimbangi oleh keputusan yang benar dari ahli lainnya. Ini membuat model lebih stabil dan akurat, serta sangat mengurangi risiko overfitting.

### 3. Bagaimana Random Forest Membuat Keputusan Akhir (Voting)?

Setelah semua pohon keputusan dilatih, ketika Random Forest menerima data baru (misalnya, data nutrisi makanan baru), setiap pohon akan membuat prediksinya sendiri (misalnya, 'sehat' atau 'tidak sehat').

Untuk mendapatkan keputusan akhir, Random Forest menggunakan mekanisme 'voting' (pemungutan suara). Untuk masalah klasifikasi seperti kita, prediksi akhir adalah kelas yang paling banyak dipilih oleh pohon-pohon individu. Jadi, jika 70 dari 100 pohon memprediksi 'Sehat' dan 30 memprediksi 'Tidak Sehat', maka Random Forest akan secara keseluruhan memprediksi 'Sehat'.

### 4. Kaitannya dengan Klasifikasi Makanan Sehat/Tidak Sehat

Dalam kasus klasifikasi makanan sehat/tidak sehat yang telah kita lakukan, Random Forest menggunakan fitur-fitur nutrisi seperti `Gula_g`, `LemakJenuh_g`, `Serat_g`, `Protein_g`, dan `Sodium_mg` sebagai input. Setiap pohon keputusan dalam 'hutan' akan menimbang fitur-fitur ini secara berbeda untuk membuat prediksinya sendiri. Misalnya:

*   Pohon 1 mungkin memprioritaskan `Serat_g` dan `Gula_g`.
*   Pohon 2 mungkin lebih fokus pada `LemakJenuh_g` dan `Sodium_mg`.

Setelah semua pohon 'memberi suara' apakah makanan baru itu 'Sehat' (1) atau 'Tidak Sehat' (0), Random Forest akan menghitung suara tersebut. Makanan akan diklasifikasikan ke dalam kategori yang menerima suara terbanyak. Probabilitas yang kita lihat (`probabilitas_baru`) adalah persentase suara untuk setiap kategori. Misalnya, `[0.39 0.61]` berarti 39% pohon memilih 'Tidak Sehat' dan 61% memilih 'Sehat'.
